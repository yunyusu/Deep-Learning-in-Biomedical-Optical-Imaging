# Gradio
Use Gradio to build a image processing tool with the feature of rotation and denoising function

```python
import gradio as gr
import cv2

# Denoise function: averaging the picture with specific size of sliding window
# input: img(image to be processed), kernel_size(the size of the sliding window)
# output: img(the processed image)
def image_gauss(img, kernel_size):
  img = cv2.fastNlMeansDenoisingColored(img, None, 10, 0, kernel_size, 20)
  return img

# Rotate function: rotate input image 90 degree clockwise
# input: img(image to be processed)
# output: img(the rotated image, or None as input is None)
def image_rotate(img):
  if img is None:
    return None
  else:
    return cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)


# initialize the output image to be the input image
def img_init(img):
  return img

with gr.Blocks() as demo:
    # The header of the image processor
    gr.Markdown("# 醫學影像處理工具(Interactive Image Proccessing Tool)")
    gr.Markdown("1. Rotate 90 degree for every click")
    gr.Markdown("2. Set the kernel size of Gaussian denoise")

    # Put the input and output image in a row to compare them easily
    with gr.Row():
      with gr.Column():
        input_image = gr.Image(label="輸入圖片", type="numpy")
        kernel_size = gr.Slider(0, 100, value=5, label="Kernel Size", interactive=True)
        rotate_btn = gr.Button("旋轉90度", variant="primary")

      with gr.Column():
        output_img = gr.Image(label="輸出結果", type="numpy")

    # Initialize the image once the input image has changed
    input_image.change(
        fn=img_init,
        inputs=input_image,
        outputs=output_img
    )

    # Update the output image as the button or the slider is changed
    rotate_btn.click(
        fn=image_rotate,
        inputs=output_img,
        outputs=output_img
    )

    kernel_size.change(
        fn=image_gauss,
        inputs=[input_image, kernel_size],
        outputs=output_img
    )

# Execute the user panel
demo.launch()
```
